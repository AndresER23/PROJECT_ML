```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
ciudades ={
    11001: 'Bogotá', 76001: 'Cali', 5001: 'Medellín', 66001: 'Pereira', 13001: 'Cartagena', 
    50001: 'Villavicencio', 8001: 'Barranquilla', 68001: 'Bucaramanga', 23001: 'Monteria', 
    73001: 'Ibague', 54001: 'Cucuta', 17001: 'Manizales', 52001: 'Pasto'
}
preguntas = {'P220':'Sexo','P5785':'Edad','P1365':'Ocupacion' }
url_personas ='https://raw.githubusercontent.com/AndresER23/PROJECT_ML/refs/heads/master/Caracteristicas%20generales%20de%20las%20personas.CSV'
df_personas = pd.read_csv(url_personas, sep=';')
df_personas= df_personas.drop(['FEX_C','ORDEN','SECUENCIA_ENCUESTA','SECUENCIA_P','NRO_ENCUESTA','P5501','P2057','P2059','P2061','P1402','P1403','P1364','P1363'], axis=1)
df_personas = df_personas.rename(columns=preguntas)
df_personas['Ocupacion'].value_counts()
ocupacion_map = {
    1: 'Trabajando',
    2: 'Buscando trabajo',
    3: 'Estudiando',
    4: 'Oficios del hogar',
    5: 'Incapacitado permanente para trabajar',
    6: 'Pensionado',
    7: 'Ocio',
    8: 'Otra actividad'
}
sexo_map = {
    1: 'Hombre',
    2: 'Mujer'}
# Reemplazar los valores en la columna 'Ocupacion'
df_personas['Ocupacion'] = df_personas['Ocupacion'].replace(ocupacion_map)
df_personas['Sexo'] = df_personas['Sexo'].replace(sexo_map)
df_personas
```

```python
url_viviendas = 'https://raw.githubusercontent.com/AndresER23/PROJECT_ML/refs/heads/master/Datos%20de%20la%20vivienda.CSV'
df_viviendas= pd.read_csv(url_viviendas, sep=';')
df_viviendas = df_viviendas.drop(['FEX_C','ORDEN','SECUENCIA_ENCUESTA','SECUENCIA_P','NRO_ENCUESTA','P1987','P1988','P1989','P1990'], axis=1)
df_viviendas['DEPMUNI'] = df_viviendas['DEPMUNI'].replace(ciudades)
df_viviendas = df_viviendas.rename(columns={'DEPMUNI':'Ciudad','P1988S1':'Estrato'})
df_viviendas = df_viviendas.drop(['CIUDADES28'], axis=1)
df_viviendas
```
